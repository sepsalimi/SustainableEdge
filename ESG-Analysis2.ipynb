```python
# Calculate the correlation matrix for numeric columns in the tech dataframe
numeric_cols = tech_df.select_dtypes('number').columns
corr = tech_df[numeric_cols].corr()

# Unstack the matrix into a Series (Variable1, Variable2) -> Correlation
corr_unstacked = corr.unstack()

# Sort by correlation value in descending order
sorted_corr = corr_unstacked.sort_values(ascending=False)

# Remove self-correlations (Variable1 == Variable2)
sorted_corr = sorted_corr[sorted_corr.index.get_level_values(0) != sorted_corr.index.get_level_values(1)]

# Convert to DataFrame for easier handling and display
corr_table = sorted_corr.reset_index()
corr_table.columns = ['Variable 1', 'Variable 2', 'Correlation']

# Remove duplicate pairs (e.g., keep (A, B) but remove (B, A))
corr_table['sorted_pair'] = corr_table.apply(lambda row: tuple(sorted((row['Variable 1'], row['Variable 2']))), axis=1)
corr_table = corr_table.drop_duplicates(subset=['sorted_pair'])
corr_table = corr_table.drop(columns=['sorted_pair'])

# Function to count non-NA pairs for two columns in the original tech_df
def count_valid_pairs(row, df):
    var1 = row['Variable 1']
    var2 = row['Variable 2']
    # Count rows where *both* columns are not NA
    valid_count = df[[var1, var2]].dropna().shape[0]
    return valid_count

# Apply the function to each row of corr_table to create the new column
corr_table['Non_NA_Count'] = corr_table.apply(lambda row: count_valid_pairs(row, tech_df), axis=1)


# --- Adjust display settings ---
import pandas as pd
pd.set_option('display.max_columns', None) 
pd.set_option('display.width', 1000) 

# Display the top N correlations (e.g., top 20)
print("Top Correlations (Highest to Lowest) with Non-NA Counts:")
# Display the table including the new column
print(corr_table.head(20)) 
```

```python
# 1. Filter by minimum sample size first
# Correlations with too few samples are unreliable regardless of value
min_samples = 30  # Common statistical threshold, but domain-specific
reliable_corr = corr_table[corr_table['Non_NA_Count'] >= min_samples].copy()

# 2. Calculate statistical significance (p-values)
def calculate_pvalue(row, df):
    var1, var2 = row['Variable 1'], row['Variable 2']
    # Get the two columns, drop rows where either has NA
    data = df[[var1, var2]].dropna()
    # Calculate pearson correlation and p-value
    r, p_value = stats.pearsonr(data[var1], data[var2])
    return p_value

# Add p-values to the filtered correlation table
reliable_corr['p_value'] = reliable_corr.apply(lambda row: calculate_pvalue(row, tech_df), axis=1)

# 3. Apply Benjamini-Hochberg correction for multiple testing (optional but recommended)
from statsmodels.stats.multitest import multipletests
reliable_corr['adjusted_p_value'] = multipletests(reliable_corr['p_value'], method='fdr_bh')[1]

# 4. Filter by statistical significance
significant_corr = reliable_corr[reliable_corr['adjusted_p_value'] < 0.05].copy()

# 5. Sort by absolute correlation strength
significant_corr['abs_corr'] = significant_corr['Correlation'].abs()
significant_corr = significant_corr.sort_values('abs_corr', ascending=False)

# 6. Check for hierarchical relationships (manual inspection needed)
# Add a column for notes
significant_corr['notes'] = ""

# 7. Display the final table of legitimate correlations
pd.set_option('display.max_rows', None)
print("Legitimate correlations (sample size ≥ {}, statistically significant):".format(min_samples))
significant_corr[['Variable 1', 'Variable 2', 'Correlation', 'Non_NA_Count', 
                 'p_value', 'adjusted_p_value', 'notes']]
```

### Visualize correlations with scatter plots

```python
# Function to create scatterplots of variable pairs
def plot_correlation_pair(var1, var2, df, ax=None):
    if ax is None:
        fig, ax = plt.subplots(figsize=(8, 6))
    
    # Get data without NAs
    data = df[[var1, var2]].dropna()
    n = len(data)
    
    # Calculate correlation
    corr_value = data[var1].corr(data[var2])
    
    # Create scatterplot
    sns.scatterplot(x=var1, y=var2, data=data, ax=ax)
    
    # Add regression line
    sns.regplot(x=var1, y=var2, data=data, scatter=False, 
                line_kws={'color': 'red'}, ax=ax)
    
    # Set title with correlation and sample size
    ax.set_title(f"Correlation: {corr_value:.3f} (n={n})")
    
    # Improve axis labels readability
    ax.set_xlabel(var1, fontsize=10)
    ax.set_ylabel(var2, fontsize=10)
    
    return ax
```
