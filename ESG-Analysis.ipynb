```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

# Display settings for better visibility

pd.set_option('display.max_columns', None)
```

```python
# Define the file path
file_path = r"C:\Users\sepeh\OneDrive\Documents\Git\SustainableEdge\SustainableEdge\SustainableEdge-ESG.csv"

# Load the CSV file
df = pd.read_csv(file_path)

# Display the first 3 rows
df.head(3)
```

```python
# Check DataFrame info
df.info()

# Check for missing values
df.isnull().sum()

# Check data types
df.dtypes
```

```python
# Select only numerical columns
numerical_df = df.select_dtypes(include=['number'])

# Display first few rows
numerical_df.head(3)
```

```python
# Cleaning column names
numerical_df.columns = numerical_df.columns.str.replace(r'\n\(FY0\)', '', regex=True).str.strip()

# Display first few rows
numerical_df.head(3)
```

```python
# Compute correlation matrix
correlation_matrix = numerical_df.corr()

# Display it
correlation_matrix
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Set figure size
plt.figure(figsize=(15,10))

# Create heatmap
sns.heatmap(correlation_matrix, annot=False, cmap='coolwarm', linewidths=0.5)

# Show the plot
plt.show()
```

```python
# Set correlation threshold
threshold = 0.8  # Adjust as needed

# Find highly correlated pairs
high_corr_pairs = correlation_matrix.abs().unstack().sort_values(ascending=False)

# Remove self-correlation (diagonal)
high_corr_pairs = high_corr_pairs[high_corr_pairs < 1]

# Display highly correlated features
high_corr_filtered = high_corr_pairs[high_corr_pairs > threshold]
print("Highly Correlated Features:")
print(high_corr_filtered)
```

```python
# Identify features related to company size
size_related_keywords = ["Employees", "Revenues", "Total Waste", "Total Energy", "Fleet", "Expenditures"]

# Compute correlation matrix
correlation_matrix = numerical_df.corr()

# Find correlations of these features with all other features
size_correlations = correlation_matrix.loc[
    correlation_matrix.index.str.contains('|'.join(size_related_keywords), case=False, na=False)
]

# Display the list of features most correlated with company size
size_correlations = size_correlations.abs().mean().sort_values(ascending=False)
print("Features highly correlated with company size:")
print(size_correlations)
```
